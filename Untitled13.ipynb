```python
import matplotlib.pyplot as plt
df['Attrition'].value_counts().plot(kind='bar')
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.show()
```

```python
df.groupby(['OverTime', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Overtime')
plt.xlabel('OverTime')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby(['Department', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby(['JobRole', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby(['JobSatisfaction', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Job Satisfaction')
plt.xlabel('Job Satisfaction')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby(['WorkLifeBalance', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Work-Life Balance')
plt.xlabel('Work-Life Balance')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby(['BusinessTravel', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Business Travel')
plt.xlabel('Business Travel')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby(['MaritalStatus', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby(['Gender', 'Attrition']).size().unstack().plot(kind='bar')

plt.title('Employee Attrition by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Employees')
plt.xticks(rotation=0)
plt.show()
```

```python
df[df['Attrition'] == 'Yes']['Age'].plot(kind='hist', bins=10)

plt.title('Age Distribution of Employees Who Left')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

```python
df.boxplot(column='MonthlyIncome', by='Attrition')

plt.title('Monthly Income by Attrition')
plt.suptitle('')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.show()
```

```python
df.boxplot(column='YearsAtCompany', by='Attrition')

plt.title('Years at Company by Attrition')
plt.suptitle('')
plt.xlabel('Attrition')
plt.ylabel('Years at Company')
plt.show()
```

```python
jobrole_attrition = (
    df.groupby('JobRole')['Attrition']
      .apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False)
)

jobrole_attrition


```

```python
department_attrition = (
    df.groupby('Department')['Attrition']
      .apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False)
)

department_attrition
```

```python
overtime_attrition = (
    df.groupby('OverTime')['Attrition']
      .apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False)
)

overtime_attrition
```

```python
satisfaction_attrition = (
    df.groupby('JobSatisfaction')['Attrition']
      .apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False)
)

satisfaction_attrition
```

```python
balance_attrition = (
    df.groupby('WorkLifeBalance')['Attrition']
      .apply(lambda x: (x == 'Yes').mean() * 100)
      .sort_values(ascending=False)
)

balance_attrition
```

```python
jobrole_attrition.plot(kind='bar')

plt.title('Attrition Rate by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=45)
plt.show()
```
